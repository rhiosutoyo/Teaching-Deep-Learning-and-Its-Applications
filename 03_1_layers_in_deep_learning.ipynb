# Layers: The building blocks of deep learning
A layer is a data processing module that takes as input one or more tensors and that outputs one or more tensors.
* Different types of layers are appropriate for different tensor formats and different types of data processing (e.g., time-series, images).
* The example belows use only Sequential models (a subclass of Model), which are simple stacks of layers, mapping a single input to a single output. Advanced model will be introduced later: Two-branch networks, Multihead networks, and Residual connections.

## 1. Dense Layer for Vector Data ((samples, features))
Simple vector data, stored in rank-2 tensors of shape (samples, features), is often processed by densely connected layers, also called fully connected or dense layers (the Dense class in Keras).

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Simulate simple vector data: 100 samples, 20 features each
vector_data = tf.random.normal((100, 20))

# Build a model for vector data
vector_model = models.Sequential([
    layers.Dense(64, activation='relu', input_shape=(20,)), # learns 64 weights for each input feature, activation='relu' adds non-linearity
    layers.Dense(10, activation='softmax')  # classification into 10 classes
])

# Forward pass (inference)
vector_output = vector_model(vector_data)
print("Dense Layer Output Shape:", vector_output.shape)
```

## 2. Conv1D or LSTM for Sequence Data ((samples, timesteps, features))
Sequence data, stored in rank-3 tensors of shape (samples, timesteps, features), is typically processed by recurrent layers, such as an LSTM layer, or 1D convolution layers (Conv1D).

### A. Using Conv1D:

```python
# Simulate sequence data: 100 samples, 50 timesteps, 16 features
sequence_data = tf.random.normal((100, 50, 16))

# Conv1D model
conv1d_model = models.Sequential([
    layers.Conv1D(32, kernel_size=3, activation='relu', input_shape=(50, 16)), # applies 32 filters/sliding windows along time steps
    layers.GlobalMaxPooling1D(), # compresses each filter into one value by taking the max
    layers.Dense(10, activation='softmax')  # classification into 10 classes
])

conv1d_output = conv1d_model(sequence_data)
print("Conv1D Output Shape:", conv1d_output.shape)
```

### B. Using LSTM:

```python
# Simulate sequence data: 100 samples, 50 timesteps, 16 features
sequence_data = tf.random.normal((100, 50, 16))

# LSTM model
lstm_model = models.Sequential([
    layers.LSTM(32, input_shape=(50, 16)),  # LSTM reads the sequence step-by-step and maintains memory (state)
    layers.Dense(10, activation='softmax')  # classification into 10 classes
])

lstm_output = lstm_model(sequence_data)
print("LSTM Output Shape:", lstm_output.shape)
```

## 3. Conv2D for Image Data ((samples, height, width, channels))
Image data, stored in rank-4 tensors, is usually processed by 2D convolution layers (Conv2D).

```python
# Simulate image data: 100 samples of 64x64 RGB images
image_data = tf.random.normal((100, 64, 64, 3))

# Conv2D model
conv2d_model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(64, 64, 3)),    # Conv2D applies filters in 2D (height x width), preserving spatial structure
    layers.MaxPooling2D((2, 2)),    # MaxPooling2D reduces the size of the feature maps
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.GlobalAveragePooling2D(),    # GlobalAveragePooling2D converts feature maps to a flat vector
    layers.Dense(10, activation='softmax')  # classification into 10 classes
])

conv2d_output = conv2d_model(image_data)
print("Conv2D Output Shape:", conv2d_output.shape)
```
